```python
print(f"Há {df.isnull().sum().sum()} valores nulos")
```

```python
Q1 = df[num_cols].quantile(0.25)
Q3 = df[num_cols].quantile(0.75)
IQR = Q3 - Q1

lim_inf = Q1 - 1.5 * IQR
lim_sup = Q3 + 1.5 * IQR

print("A porcentagem de outliers por coluna de forma descendente é: ")
porcentagens_outliers = df[(df[num_cols] < lim_inf) | (df[num_cols] > lim_sup)].count()/df[num_cols].count()*100
porcentagens_outliers.sort_values(ascending=False)
```

```python
print(f"Há {df.duplicated().sum()} valores duplicadas")
```

### <font color='cyan'> Q19: Calcule e exiba o Coeficiente de Correlação de Pearson ($r$) entre elas.</font>

```python
import seaborn as sns
```

```python
corr = df["flow_duration"].corr(df["flow_pkts_payload.tot"])

sns.scatterplot(data=df, x="flow_duration", y="flow_pkts_payload.tot")
plt.title(f"flow_duration vs flow_pkts_payload.tot | Pearson: {corr}")
plt.show()
```

### <font color='cyan'> Q20: Interprete o coeficiente de correlação, em termos de sinal e magnitude.</font>

O sinal da correlação é positivo (r = 0,07), o que significa que o total de bytes
de payload dos pacotes do fluxo tende a aumentar quando a duração do fluxo aumenta.
No entanto, a magnitude do coeficiente é tão baixa que a relação entre as duas
variáveis é desprezível e não deve ser levada em consideração.
